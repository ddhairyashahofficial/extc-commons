```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
# Ensure 'netflix_revenue_updated.csv' is uploaded to your Colab files
df = pd.read_csv('netflix_revenue_updated.csv')

# Clean column names (remove potential trailing spaces)
df.columns = df.columns.str.strip()

# 2. Create the Basic Scatter Plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Netflix Streaming Memberships', y='Global Revenue', s=100, color='blue')
plt.title('Scatter Plot: Global Revenue vs. Netflix Memberships', fontsize=16)
plt.xlabel('Netflix Streaming Memberships', fontsize=12)
plt.ylabel('Global Revenue', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# 3. Create the Scatter Plot with a Regression Line
plt.figure(figsize=(10, 6))
sns.regplot(data=df, x='Netflix Streaming Memberships', y='Global Revenue',
            scatter_kws={'s': 100, 'color': 'blue'},
            line_kws={'color': 'red', 'linewidth': 2})
plt.title('Global Revenue vs. Memberships with Regression Line', fontsize=16)
plt.xlabel('Netflix Streaming Memberships', fontsize=12)
plt.ylabel('Global Revenue', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Optional: Print correlation coefficient
correlation = df['Netflix Streaming Memberships'].corr(df['Global Revenue'])
print(f"Correlation Coefficient: {correlation:.4f}")
```
