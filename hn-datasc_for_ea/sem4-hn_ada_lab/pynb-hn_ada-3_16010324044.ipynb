```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import datetime as dt
```

```python
dates = pd.date_range(start="2024-01-01", periods=100, freq="D")
temperature = np.random.randint(20, 35, size=100)

df = pd.DataFrame({
    "Date": dates,
    "Temperature": temperature
})

df.head()
```

```python
df.to_csv("temperature.csv", index=False)

ts = pd.read_csv("temperature.csv")
ts.head()
```

```python
ts["Date"] = pd.to_datetime(ts["Date"])
ts.set_index("Date", inplace=True)

ts.head()
```

```python
ts.plot()
plt.title("Daily Temperature Time Series")
plt.show()
```

```python
now = dt.datetime.now()
print("Current Date Time:", now)
```

```python
today = dt.date.today()
future = today + dt.timedelta(days=10)

print("Today:", today)
print("After 10 days:", future)
```

```python
ts["Year"] = ts.index.year
ts["Month"] = ts.index.month
ts["Day"] = ts.index.day

ts.head()
```

```python
monthly_avg = ts.resample("M").mean()
monthly_avg
```
