```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
data = {
    "Age": [22, 25, np.nan, 28, 35, np.nan, 40],
    "Salary": [25000, np.nan, 32000, 28000, np.nan, 50000, 52000],
    "Experience": [1, 3, 5, np.nan, 10, 12, np.nan]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df_mean = df.fillna(df.mean())
df_mean
```

```python
df_median = df.fillna(df.median())
df_median
```

```python
df_ffill = df.fillna(method="ffill")
df_ffill
```

```python
imputer_mean = SimpleImputer(strategy="mean")
df_sklearn_mean = pd.DataFrame(
    imputer_mean.fit_transform(df),
    columns=df.columns
)

df_sklearn_mean
```

```python
imputer_median = SimpleImputer(strategy="median")

df_sklearn_median = pd.DataFrame(
    imputer_median.fit_transform(df),
    columns=df.columns
)

df_sklearn_median
```
